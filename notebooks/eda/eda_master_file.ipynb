# 3. Processing

```python
df_master_stats_proc = df_master_stats.copy()
df_master_stats_proc = df_master_stats_proc.reset_index()

#Rename N Number to Tail number
df_master_stats_proc.rename(columns={'N-NUMBER': 'TailNumber'}, inplace=True)
```

```python

list_cols = [    
  'TailNumber',
  'YEAR MFR',
  'LAST ACTION DATE',
  'CERT ISSUE DATE',
  'AIR WORTH DATE',
  'EXPIRATION DATE',
  "MFR MDL CODE"
]
```

```python
df_master_stats_proc= df_master_stats_proc.loc[:, list_cols]
```

```python
#Data conversion.
#TODO: Keeping na as such. Do we need to exclude those records or change it to default date?
#YEAR MFR is still in float due to na values. Once default is assigned then it will becomes int.

# Convert 'CERT ISSUE DATE' from float to datetime format (YYYYMMDD). 
df_master_stats_proc['CERT ISSUE DATE'] = df_master_stats_proc['CERT ISSUE DATE'].fillna(0).apply(lambda x: datetime.strptime(str(int(x)), '%Y%m%d') if x != 0 else None)

# Convert 'AIR WORTH DATE' from float to datetime format (YYYYMMDD).
df_master_stats_proc['AIR WORTH DATE'] = df_master_stats_proc['AIR WORTH DATE'].fillna(0).apply(lambda x: datetime.strptime(str(int(x)), '%Y%m%d') if x != 0 else None)

# Convert 'EXPIRATION DATE' from float to datetime format (YYYYMMDD).
df_master_stats_proc['EXPIRATION DATE'] = df_master_stats_proc['EXPIRATION DATE'].fillna(0).apply(lambda x: datetime.strptime(str(int(x)), '%Y%m%d') if x != 0 else None)

# Convert 'LAST ACTION DATE' from float to datetime format (YYYYMMDD).
df_master_stats_proc['LAST ACTION DATE'] = df_master_stats_proc['LAST ACTION DATE'].fillna(0).apply(lambda x: datetime.strptime(str(int(x)), '%Y%m%d') if x != 0 else None)

# Convert 'YEAR MFR' from float to integer.
df_master_stats_proc['YEAR MFR'] = df_master_stats_proc['YEAR MFR'].apply(lambda x: int(x) if pd.notna(x) else None)

```

```python
# Calculate NaN statistics
total_records = len(df_master_stats_proc)
nan_count = df_master_stats_proc.isna().sum()
nan_percentage = (nan_count / total_records) * 100

# Display results
nan_stats = pd.DataFrame({'NaN Count': nan_count, 'Total Records': total_records, 'NaN Percentage (%)': nan_percentage}).round(2)
print(nan_stats)
```

```python
#Derived columns
current_year = datetime.now().year
current_date = datetime.now()

# Calculate age and create a new column 'Age_Year_Mfr'
df_master_stats_proc['Age_Year_Mfr'] = df_master_stats_proc['YEAR MFR'].apply(lambda x: current_year - x if pd.notna(x) else None)

# Calculate age in days for last action, cert issue and Air worth dates
df_master_stats_proc['AgeDays_Last_Action_Date'] = df_master_stats_proc['LAST ACTION DATE'].apply(lambda x: (current_date - x).days if pd.notna(x) else None)
df_master_stats_proc['AgeDays_Cert_Issue_Date'] = df_master_stats_proc['CERT ISSUE DATE'].apply(lambda x: (current_date - x).days if pd.notna(x) else None)
df_master_stats_proc['AgeDays_Air_Worth_Date'] = df_master_stats_proc['AIR WORTH DATE'].apply(lambda x: (current_date - x).days if pd.notna(x) else None)

# Calculate days until EXPIRATION DATE
df_master_stats_proc['Days_Until_Expiration_Date'] = df_master_stats_proc['EXPIRATION DATE'].apply( lambda x: (x - current_date).days if pd.notna(x) else None)
```
